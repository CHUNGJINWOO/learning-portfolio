```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
t = np.arange(0, 6, 0.01)
t
```

```python
t.shape
```

```python
y = np.sin(np.pi * t)
y.shape
```

```python
plt.figure(figsize=(12,6))
plt.plot(t, y)
plt.grid()
plt.show()           
```

```python
t1 = np.arange(0, 3, 0.01)
y1 = np.sin(np.pi * 5 * t1)

plt.figure(figsize=(12,6))
plt.plot(t1, y1)
plt.grid()
plt.show() 
```

```python
def draw_sin(t, A, f, b):
    y = A * np.sin(np.pi * f * t) + b

    plt.figure(figsize=(12,6))
    plt.plot(t, y)
    plt.grid()
    plt.show() 
```

```python
t = np.arange(0, 3, 0.01)
draw_sin(t, 1, 1, 0)
```
