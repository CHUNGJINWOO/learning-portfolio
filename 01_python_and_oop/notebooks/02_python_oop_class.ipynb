```python
class DrawSin:
    def __init__(self, amp, freq, bias, end_time) :
        self.amp = amp
        self.freq = freq
        self.bias = bias
        self.end_time = end_time

    def calc_sin(self):
        self.t = np.arange(0, self.end_time, 0.01)
        return self.amp * np.sin(2*np.pi*self.freq*self.t) + self.bias
    def draw_sin(self):
        y = self.calc_sin()
        plt.figure(figsize=(12,6))
        plt.plot(self.t, y)
        plt.grid()
        plt.show()
```

```python
ds = DrawSin(1, 1, 0, 3)
ds.draw_sin()
```

```python
ds.freq = 2
ds.draw_sin()
```

```python
class DrawSinusoidal(DrawSin): #class 상속
    def calc_cos(self):
        self.t = np.arange(0, self.end_time, 0.01)
        return self.amp * np.cos(2*np.pi*self.freq*self.t) + self.bias

    def draw_cos(self):
        y = self.calc_cos()
        plt.figure(figsize=(12,6))
        plt.plot(self.t, y)
        plt.grid()
        plt.show()
```

```python
class DrawSinusoidal2(DrawSinusoidal):
    def draw_sin(self):
        y = self.calc_sin()
        plt.figure(figsize=(12,6))
        plt.plot(self.t, y)
        plt.title('Sin Graph ')
        plt.ylabel('Sin')
        plt.xlabel('time (sec)')
        plt.grid()
        plt.show()
```

```python
class DrawSinusoidal3(DrawSinusoidal2):
    def __init__(self, amp, freq, bias, end_time, ts):
        super().__init__(amp, freq, bias, end_time)
        self.ts = ts

    def calc_sin(self):
        self.t = np.arange(0, self.end_time, self.ts)
        return self.amp * np.sin(2*np.pi*self.freq*self.t) + self.bias

    def draw_sin(self):
        y = self.calc_sin()
        plt.figure(figsize=(12,6))
        plt.plot(self.t, y)
        plt.title('Sin Graph ')
        plt.ylabel('Sin')
        plt.xlabel('time (sec)')
        plt.grid()
        plt.show()
```
